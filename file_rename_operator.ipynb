```python
import os
import random

def rename_randomized_images(folder_path, prefix):
    # 폴더 내 파일 목록 불러오기
    files = [f for f in os.listdir(folder_path) 
             if os.path.isfile(os.path.join(folder_path, f))]

    # 파일 순서 랜덤 셔플
    random.shuffle(files)

    # 파일명 재부여
    for idx, filename in enumerate(files, start=1):
        ext = os.path.splitext(filename)[1]  # 확장자 유지
        new_name = f"{prefix}_{idx:06d}{ext}"  # 예: acc_000001.jpg
        os.rename(
            os.path.join(folder_path, filename),
            os.path.join(folder_path, new_name)
        )

    print(f"Renaming completed in: {folder_path}")

# 사용 예시
#rename_randomized_images(r"C:\Users\icebl\OneDrive\PSYJIN\0. 건국대 미래국방기술융합학과\3학기 교과목(2026년 3월-)\1. AI프로그래밍입문(안민호 교수님)\기말과제\traffic_context_accident\selected_600\accident", "acc")
rename_randomized_images(r"C:\Users\icebl\OneDrive\PSYJIN\0. 건국대 미래국방기술융합학과\3학기 교과목(2026년 3월-)\1. AI프로그래밍입문(안민호 교수님)\기말과제\traffic_context_accident\selected_600\non_accident", "non")
```
